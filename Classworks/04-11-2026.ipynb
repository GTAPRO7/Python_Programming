```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#Pandas have 2 datastructres - Dataframe and Series

iris = pd.read_excel(r"C:\Users\gokul\OneDrive\Desktop\MCA\SEM 2\Python\Classworks\iris.xlsx")
iris.head()
```

```python
print(iris.groupby('Species').size())
print(iris.groupby(['Species']).min())
```

```python
setosa = iris[iris['Species'] == 'Iris-setosa']
versicolor = iris[iris['Species'] == 'Iris-versicolor']
virginica = iris[iris['Species'] == 'Iris-virginica']
```

```python
setosa['SepalWidth'].mean()
```

```python
#Simple Plots
#Histogram

data = setosa["SepalWidth"]
data.plot(kind = "hist", label = "setosa", color = 'g', fontsize = 10)
plt.title('Setosa SepalWidth - Histogram')
plt.xlabel('SepalWidthCm')
plt.show()
```

```python
#Scatter Plot

setosa.plot(x="SepalLength", y="SepalWidth", kind="scatter", label='setosa', color='g')
plt.title('Scatter Plot - Setosa')
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.show()
```

```python
#Line Plot

iris["SepalWidth"].plot(kind = "line",
                        color = 'black', linewidth = '1.5',
                        ls = ':', legend = 'setosa')
plt.axis([0, 150, 2, 5])
plt.title('Line Plot')
plt.xlabel('SepalWidthCm')
plt.show()
```

```python
#Line Plot - Species Wise

setosa["SepalWidth"].plot(kind = "line",
                        color = 'r', linewidth = '1.5',
                        ls = '-')
versicolor["SepalWidth"].plot(kind = "line",
                        color = 'g', linewidth = '1.5',
                        ls = '-')
virginica["SepalWidth"].plot(kind = "line",
                        color = 'b', linewidth = '1.5',
                        ls = '-')

plt.title('Line Plot')
plt.xlabel('SepalWidthCm')
plt.legend(['setosa','versicolor','virginica'])
plt.show()
```

```python
#Area Plot

setosa.plot.area(stacked = False)
plt.show()
```

```python
#Area Plot

setosa.plot.area(stacked = True)
plt.show()
```

```python
setosa.head()
```

```python
fig, axs = plt.subplots(1, 2, sharey = False, tight_layout = True)

axs[0].hist(setosa['SepalWidth'], bins = 5)
axs[1].hist(versicolor['SepalWidth'])
```

```python
x = np.linspace(0, 50, 50)
fig, ax = plt.subplots(figsize = (5, 2.7), layout = 'constrained')
ax.plot(x, setosa.SepalLength)
ax.plot(x, versicolor.SepalLength)
ax.plot(x, virginica.SepalLength)
ax.set_xlabel('Observations')
ax.set_ylabel('SepalLengthCm')
ax.set_title("Line Plot-SepalLengthCm")
ax.legend(['setosa','versicolor','virginica'])
plt.show()
```

```python
#Bargraph

Type = list(iris['Species'].unique())
count = list(iris['Species'].value_counts())
Avg = list(iris.groupby('Species')["SepalLength"].mean())
plt.bar(Type, Avg, color = ['maroon', 'yellow', 'blue'], width = 0.4)
#.barh--------height=0.4
plt.title('Bar Plot')
plt.xlabel('SepalLengthCm')
plt.ylabel('AverageSepalLengthCm')
plt.show()
```

```python
#Pie Chart

e = (0.1, 0.1, 0.1)
Avg = list(iris.groupby('Species')["SepalLength"].mean())
plt.pie(Avg, explode = e, labels = ['setosa', 'versicolor', 'virginica'])
plt.title("Pie Chart")
plt.legend(title = "Iris Flowers:")
plt.show()
```

```python
#Box Plot

setosa.plot.box()

#notch = 'True', vert = 0

plt.title("Box Plot")
plt.legend(labels = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm'],
           title = "Iris Flowers:")
plt.show()
```
