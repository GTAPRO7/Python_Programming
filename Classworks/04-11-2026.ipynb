```python
import pandas as pd
import matplotlib.pyplot as plt
#Pandas have 2 datastructres - Dataframe and Series

iris = pd.read_excel(r"C:\Users\gokul\OneDrive\Desktop\MCA\SEM 2\Python\Classworks\iris.xlsx")
iris.head()
```

```python
print(iris.groupby('Species').size())
print(iris.groupby(['Species']).min())
```

```python
setosa = iris[iris['Species'] == 'Iris-setosa']
versicolor = iris[iris['Species'] == 'Iris-versicolor']
virginica = iris[iris['Species'] == 'Iris-virginica']
```

```python
setosa['SepalWidth'].mean()
```

```python
#Simple Plots
#Histogram

data = setosa["SepalWidth"]
data.plot(kind = "hist", label = "setosa", color = 'g', fontsize = 10)
plt.title('Setosa SepalWidth - Histogram')
plt.xlabel('SepalWidthCm')
plt.show()
```

```python
#Scatter Plot

setosa.plot(x="SepalLength", y="SepalWidth", kind="scatter", label='setosa', color='g')
plt.title('Scatter Plot - Setosa')
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.show()
```

```python
#Line Plot

iris["SepalWidth"].plot(kind = "line",
                        color = 'black', linewidth = '1.5',
                        ls = ':', legend = 'setosa')
plt.axis([0, 150, 2, 5])
plt.title('Line Plot')
plt.xlabel('SepalWidthCm')
plt.show()
```

```python
#Line Plot - Species Wise

setosa["SepalWidth"].plot(kind = "line",
                        color = 'r', linewidth = '1.5',
                        ls = '-')
versicolor["SepalWidth"].plot(kind = "line",
                        color = 'g', linewidth = '1.5',
                        ls = '-')
virginica["SepalWidth"].plot(kind = "line",
                        color = 'b', linewidth = '1.5',
                        ls = '-')

plt.title('Line Plot')
plt.xlabel('SepalWidthCm')
plt.legend(['setosa','versicolor','virginica'])
plt.show()
```

```python
#Area Plot

setosa.plot.area(stacked = False)
plt.show()
```
