```python
x = input("Please enter the first word :")
y = input("Please enter the second word :")
print("First Word :", x)
print("Second Word :", y)
print("Performing Concatenation...........")
print("The word after Concatenation :", x+y)

print("Printing the firts word three times :", x*3)
print("Performing Slicing...........")
print("Slicing :", x[0:2])
print("Slicing :", x[3:6])
print("Slicing :", y[-6:-2])
print()
```

```python
x = (1,2,3,4,5,6,7)
y = ("apple","mango","jackfruit")
z = (1,2,3,4,5,6,7,"apple","mango","jackfruit")
print("Numbers only Tuple :", x)
print("Characters only Tuple :", y)
print("Mixed Tuple :", z)
print(y[1])
print(x[1]*3)
print(z[0:3])
print(z[-3:-1])
print(z[-5:])
##z[1] = 3 'tuple' object does not support item assignment
print(len(z))
print(z.count(7))
print(z.index(7))
```
