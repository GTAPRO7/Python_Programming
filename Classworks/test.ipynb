```python
print("Celsius to Fahrenheit Conversion")
cel = float (input("Please enter the Celsius :"))
fht = ((cel * 9/5) + 35)
forcel = f"{cel:.2f}"
forfht = f"{fht:.2f}"
print("Celsius =", forcel)
print("Converted Fahrenheit =",forfht)
```

```python
print("Fahrenheit to Celsius Conversion")
fht = float (input("Please enter the Fahrenheit :"))
cel = ((fht - 35) * 5/9)
forfht = f"{fht:.2f}"
forcel = f"{cel:.2f}"
print("Fahrenheit =", forfht)
print("Converted Celsius =",forcel)
```

```python
##John deposited an initial sum of Rs.3000 in his account. After 3 years, the balance reaches Rs.3335.8 due to composite interest.
##What is the interest rate per annual?

IniSum = int (input("Please enter the initial sum :"))
Yrs = int (input("Please enter the number of years :"))
bln = float (input("Please enter the balance due to composite interest :"))
IntRt = ((((bln/IniSum) ** (1/Yrs)) - 1) * 100)
forIntRt = f"{IntRt:.2f}"
print("Interest Rate =", forIntRt)
```
