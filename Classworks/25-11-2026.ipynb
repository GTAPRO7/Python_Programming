```python
import sqlite3
```

```python
conn = sqlite3.connect("students1.db")
cur = conn.cursor()
```

```python
cur.execute("CREATE TABLE IF NOT EXISTS students1 (" \
            "id INT PRIMARY KEY, name TEXT, gpa REAL)")
```

```python
cur.execute("INSERT INTO students1 VALUES (?, ?, ?)",
            (101, "Aarav", 3.8))
conn.commit()
```

```python
cur.execute("SELECT * FROM students1 WHERE gpa >= ?", (3.5,))
print("Read:", cur.fetchall())
```

```python
cur.execute("UPDATE students1 SET gpa = ? WHERE id = ?",
            (3.9, 101))
conn.commit()

cur.execute("SELECT * FROM students1 WHERE id = ?", (101,))
print("After Update", cur.fetchone())
```

```python
cur.execute("DELETE FROM students1 WHERE id = ?", (101,))
conn.commit()
print("After Delete", cur.fetchall())
conn.close()
```
