```python
import pandas as pd

matches = pd.read_excel(r"C:\Users\gokul\OneDrive\Desktop\MCA\SEM 2\Python\Classworks\Match_Info.xlsx")
matches.shape
matches.head()
```

```python
deliveries = pd.read_excel(r"C:\Users\gokul\OneDrive\Desktop\MCA\SEM 2\Python\Classworks\Ball_By_Ball_Match_Data.xlsx")
deliveries.shape
deliveries.head()
```

```python
matches_analysis = matches.copy()

#Data Conversion 
matches_analysis['match_date'] = pd.to_datetime(matches_analysis['match_date'], errors='coerce')
matches_analysis['season'] = matches_analysis['match_date'].dt.year

print('Season range:', matches_analysis['season'].min(), 'to', matches_analysis['season'].max())
display(matches_analysis[['match_number', 'match_date', 'season']].head())
print('\n--- Matches Dataset Summary ---')
matches_analysis.info()
```

```python
deliveries_analysis = deliveries.copy()
```

```python
numeric_Delivery_colomns = [
    'ID', 'Innings' , 'Overs','BallNumber' , 'BatsmanRun',
    'ExtrasRun','TotalRun','IsWicketDelivery'
]

for col in numeric_Delivery_colomns:
    deliveries_analysis[col] = pd.to_numeric(deliveries_analysis[col], errors='coerce')

deliveries_analysis.info()
```

```python
season_match_count = (
    matches_analysis.groupby('season')['match_number']
    .nunique()
    .reset_index(name = 'Number_of_Matches')
    .sort_values('season')
)

display(season_match_count)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (12, 5))
plt.plot(
    season_match_count['season'],
    season_match_count['Number_of_Matches'],
    marker = '*',
    linewidth = 2,
    color = 'blue'
)

plt.title('Number of IPL Matches Played Season-wise', fontsize=14, fontweight='bold')
plt.xlabel('Season', fontsize = 12)
plt.xticks(season_match_count['season'],rotation=45)
plt.grid(True, linestyle = '--', alpha = 0.7)
plt.tight_layout()
plt.show()
```

```python
#Calculate first-innings scores for each match

first_innings_by_match = (
    deliveries_analysis[deliveries_analysis["Innings"] == 1]
    .groupby("ID")["TotalRun"]
    .sum()
    .reset_index(name = "First_Innings_Total")
)

#Merge Season Information

season_score_data = first_innings_by_match.merge(
    matches_analysis[["match_number", "season"]],
    left_on  = "ID",
    right_on = "match_number",
    how = "left"
)

season_score_data = season_score_data.dropna(subset="season")
season_score_data
```

```python
#Arrange data season-wise

seasons = sorted(season_score_data["season"].unique())

score_groups = [
    season_score_data.loc[
        season_score_data["season"] == season,
        "First_Innings_Total"
    ]
    for season in seasons
]
score_groups
```

```python
#Create combined violin and box plot

fig, ax = plt.subplots(figsize = (14, 7))

#Violin Plot

ax.violinplot(
    score_groups,
    showmeans=False,
    showmedians=False,
    showextrema=False
)

#Box plot overlay

ax.boxplot(
    score_groups,
    widths=0.15,
    patch_artist=False,
    showfliers=True
)

ax.set_title("Combined Violin and Box Plot of firs-innings scores by season")
ax.set_xlabel("IPL Seasons")
ax.set_ylabel("First-Innings Total")

ax.set_xticks(
    range(1, len(seasons) + 1),
    seasons,
    rotation = 45
)

ax.grid(axis = "y", alpha = 0.3)
plt.tight_layout()
plt.show()
```

```python
toss_by_season = pd.crosstab(
    matches_analysis["season"],
    matches_analysis["toss_decision"]
)

display(toss_by_season)
toss_by_season.plot(
    kind = "bar",
    figsize = (12, 6)
)

plt.title("Toss Decisons by Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.xticks(rotation = 45)
plt.legend(title = "Toss Decison")
plt.grid(axis = "y", alpha = 0.3)
plt.show()
```

```python
batter_runs = (
    deliveries_analysis.groupby('Batter')['BatsmanRun']
    .sum()
    .sort_values(ascending = False)
    .head(10)
    .rename('Total_Batter_Runs')
    .reset_index()    
)

display(batter_runs)

plt.figure(figsize = (10, 6))

plt.barh(
    batter_runs['Batter'][::-1],
    batter_runs['Total_Batter_Runs'][::-1],
    color = 'green'
)

```

```python
toss_analysis = matches_analysis.dropna(subset = ['toss_winner', 'toss_decision', 'winner']).copy()
toss_analysis['Toss_Winner_Won_Match'] = toss_analysis['toss_winner'] == toss_analysis['winner']
display(toss_analysis[['match_number', 'toss_winner', 'toss_decision', 'winner', 'Toss_Winner_Won_Match']].head())
```

```python
toss_summary = (
    toss_analysis.groupby('toss_decision')
    .agg(
        Tosses = ('match_number', 'count'),
        Matches_won_By_Toss_Winner = ('Toss_Winner_Won_Match', 'sum')
    )
)

toss_summary['Win_Percentage'] = (
    toss_summary['Matches_Won_By_Toss_Winner'] / toss_summary['Tosses'] * 100
).round(2)
display(toss_summary)
toss_summary[['Win_Percentage']].plot(kind = 'bar', figsize = (7,4), legend = False, color = 'blue')
plt.title('Matches-Win Percentage by Toss Decision', fontsize = 14, fontweight = 'bold')
plt.xlabel('Toss Decison', fontsize = 12)
plt.ylabel('Win Percentage (%)', fontsize = 12)
plt.xticks(rotation = 45)
plt.legend(title = "Toss Decison")
plt.grid(axis = "y", alpha = 0.3)
plt.show()

```
